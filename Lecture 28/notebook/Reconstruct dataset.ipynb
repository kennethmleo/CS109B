```python
import numpy as np
from tensorflow.keras.datasets import mnist

(x_train, _), (x_test, _) = mnist.load_data()

x_train = x_train.astype("float32")
x_test = x_test.astype("float32")

x_train_mini = x_train[:10000]   # First 10,000 training images
x_test_mini = x_test[:2000]      # First 2,000 test images

np.save('mnist_mini_train.npy', x_train_mini)
np.save('mnist_mini_test.npy', x_test_mini)

print("Files created: mnist_mini_train.npy, mnist_mini_test.npy")
```
